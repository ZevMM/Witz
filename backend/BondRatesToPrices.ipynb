```python
class Bond:
  def __init__(self, face_value, coupon_rate, maturity):
    self.face_value = face_value
    self.coupon_rate = coupon_rate
    self.steps = maturity * 2

  def coupon_payment(self):
    if self.steps == 0:
      return (1 + self.coupon_rate) * self.face_value
    return self.coupon_rate * self.face_value

  def present_value(self, discount_rate):
    x =  self.coupon_rate * self.face_value
    y = [x / ((1 + discount_rate) ** i) for i in range(1, self.steps + 1)] + [self.face_value / ((1 + discount_rate) ** self.steps)]
    return sum(y)

  def __repr__(self):
    return f"<FV: {self.face_value}, CR: {self.coupon_rate}, S: {self.steps}>"


def generatePriceSeries(initial, steps, maturity, rates):
  prices = [initial.present_value(rates[0])]
  holdings = [initial]

  for i in range(1, steps + 1):
    v = 0
    c = 0
    for h in holdings:
      v += h.present_value(rates[i])
      c += h.coupon_payment()
      h.steps -= 1
    prices.append(v)
    holdings.append(Bond(c, rates[i], maturity))
    holdings = holdings[:maturity * 2]

  return prices

rates = [0.02527, 0.0197, 0.02121, 0.02176, 0.02202, 0.02389, 0.02015, 0.01671, 0.01351, 0.01266, 0.01407, 0.01409, 0.01198, 0.01452, 0.01451, 0.0164, 0.01573, 0.01646, 0.01858, 0.02182, 0.02425, 0.02301, 0.02263, 0.02065, 0.01897, 0.01927, 0.02092, 0.01942, 0.01787, 0.01905, 0.02098, 0.02182, 0.02448, 0.02947, 0.03057, 0.03119, 0.02974, 0.03255, 0.03765, 0.04204, 0.03822, 0.03975, 0.03661, 0.03931, 0.03689, 0.03678, 0.03856, 0.03855, 0.04019, 0.04204, 0.04711, 0.05022, 0.04513, 0.04019, 0.04215, 0.04375, 0.04349, 0.0479, 0.04718, 0.04718]

myBond = Bond(100, 0.05, 30)

print(generatePriceSeries(myBond, 59, 30 , rates))

#do I wanna use a different discount rate for the shorter bonds? Nah, just let that be a slight inaccuracy. To be totally accurate, I should use the yield curve so bonds with shorter maturity are worth more (b/c they're being compared to a shorter maturity)
```

```python
stockprice = [100, 120, 130, 150, 143, 118, 95, 92, 100]
cash = 100
loans = 200
shares = 3
leverage = 3

value = []

for s in range(len(stockprice)):
  # 1. determine how much money we have
  stock_val = stockprice[s] * shares
  cash = stock_val - loans

  # 2. determine how many more shares to buy/sell
  newloanamt = cash * (leverage - 1)
  sharesToBuy = (newloanamt - loans) / stockprice[s]

  # 3. update parameters
  loans = newloanamt
  shares += sharesToBuy

  print(cash, loans, shares)
  print(cash + loans, stockprice[s] * shares)
  value.append(cash)


print(value)


#We're just altering the amount of stock to keep cash to loan ratio constant.
#price goes up, take out a loan and buy more. Price goes down, sell some and repay loans.
```
